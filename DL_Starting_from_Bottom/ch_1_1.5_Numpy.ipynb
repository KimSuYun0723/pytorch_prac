# 1.5 Numpy

### Load Numpy

```python
import numpy as np
```

### Numpy array

```python
x = np.array([1,2,3])
print(x)
type(x)
```

```python
x = np.array([1,2,3])
y = np.array([2,4,6])
print(x+y, x-y, x/y) # Broadcast
```

### N-dimension array

```python
A = np.array([[1,2], [3,4]])
print(A)

print(A.shape) # array's shape(형상)
print(A.dtype) # array's data type

B = np.array([[3,0], [0,6]])
print(A+B)
print(A*B)
```

### Broadcast

- 1D array == vector
- 2D array == matrix
- what generalize vector&matrix == tensor

```python
A = np.array([[1,2], [3,4]])
B = np.array([10,20]) #expand to 2by2 matrix
A*B
```

### Approach elements

```python
X = np.array([[51,55], [14,19], [0,4]])
X = X.flatten() # change to 1D array (평탄화)
print(X)

X[np.array([0,2,4])]
```

# 1.6 matplotlib

### Simple Graph

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(0,6,0.1) # from 0 to 6, gap 0.1
y = np.sin(x)

plt.plot(x,y)
plt.show()
```

### Function of pyplot

```python
import numpy as np
import matplotlib.pyplot as plt

# prepare data
x = np.arange(0,6,0.1)
y1 = np.sin(x)
y2 = np.cos(x)

# draw graph
plt.plot(x,y1,label="sin")
plt.plot(x,y2,linestyle="--",label="cos") 
plt.xlabel("x")
plt.ylabel("y")
plt.title("sin & cos")
plt.legend()
plt.show()
```
